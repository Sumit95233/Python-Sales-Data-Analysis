## Sales Analyzer

### Reading the files

```python
file = open("sales_data_public.csv")
lines = file.readlines()
print(lines)
```

```python
newData = []
for line in lines[1: ]:
    field= line.strip().split(",")
    field[5]= int(field[5])
    field[6]= int(field[6])
    newData.append(field)

    print(newData)
```

```python
# Q(1) What is the total revenue from all sales ?

total_revenue = 0
for row in newData:
    quantity = row[5]
    price = row[6]
    total_revenue += quantity*price

print("Total_Revenue: ₹",total_revenue)
```

```python
# Q(2) How many total orders were placed ?

total_orders= len(newData)
print(total_orders)
```

```python
# Q(3) How many total items(quantity) were sold ?

total_quantity = 0
for row in newData:
    total_quantity += row[5]

print("Total_Quantity_Sold",total_quantity)
```

```python
# Q(4) What is the total revenue generated by the each category ?

category_revenue={}

for row in newData:
    category=row[4]
    revenue=row[5]*row[6]

    if category in category_revenue:
        category_revenue[category] += revenue
    else:
        category_revenue[category] = revenue

for cat,rev in category_revenue.items():
    print(f"{cat}: ₹{rev}")
```

```python
# Q(5) What is the total revenue generated by each product ?

product_revenue = {}

for row in newData:
    product=row[3]
    revenue=row[5]*row[6]

    if product in product_revenue:
        product_revenue[product] += revenue
    else:
        product_revenue[product] = revenue

for pro,rev in product_revenue.items():
    print(f"{pro}: ₹{rev}")

```

```python
# Q(6) Which product has the highest quantity sold ? 

product_quantity = {}

for row in newData:
    product=row[3]
    quantity=row[5]

    if product in product_quantity:
        product_quantity[product] += quantity
    else:
        product_quantity[product] = quantity

highest_Sold = None
max_Quantity = 0

for product,qty in product_quantity.items():
    if qty > max_Quantity:
        max_Quantity = qty
        highest_Sold = product

print("Highest_Sold_product :",highest_Sold, "\n Max_Quantity :", max_Quantity)
```

```python
# Q(7) Which customer spent the most money ?

customer_revenue = {}

for row in newData:
    customer=row[2]
    revenue=row[5]*row[6]

    if customer in customer_revenue:
       customer_revenue[customer] += revenue
    else:
       customer_revenue[customer] = revenue

Customer_name = None
max_revenue_spent = 0

for customer,rev in customer_revenue.items():
    if rev > max_revenue_spent:
        Customer_name= customer
        max_revenue_spent = rev

print("Top customer is  :",Customer_name,"\n Max revenue spent is : ₹",max_revenue_spent)



```

```python
# Q(8) What is the total revenue for each date ? 

daily_revenue = {}

for row in newData:
    date = row[1]
    revenue=row[5]*row[6]

    if date in daily_revenue:
        daily_revenue[date] += revenue
    else:
        daily_revenue[date] = revenue

print("Daily revenue : ",daily_revenue)

for date,rev in daily_revenue.items():
    print(f"{date}: ₹{rev}")


```

```python
# Q(9) What is the total revenue between two given dates ?

from datetime import datetime
start = input("Enter Start Date(YYYY-MM-DD)")
end = input("Enter End Date(YYYY-MM-DD)")

start_date = datetime.strptime(start,"%Y-%m-%d")
end_date = datetime.strptime(end,"%Y-%m-%d")

total_dateRev = 0

for row in newData:
    order_date = datetime.strptime(row[1],"%Y-%m-%d")

    if start_date <= order_date <= end_date:
        total_dateRev += row[5]*row[6]

print("Revenue From ",start,"and",end,"is ₹",total_dateRev)
```

```python
# Q(10) Export all orders under the category "Electronics" category into a new CSV file. 

elecfile = open("electricals_public.csv", "w")
elecfile.write("OrderID,Date,Customer,Product,Category,Quantity,Price\n")
for row in newData:
    if row[4] == "Electronics":
        row_str = ",".join([str(item) for item in row])
        elecfile.write(row_str + "\n")
elecfile.close()
print("All electronics items saved as electricals_public.csv")
```
